```python
import numpy as np
import pandas as pd
```

```python
data={'Name':['Amit','Bebo','Chetan','Amit','Divya','Ravi','Sneha','Karan',
              'Pooja','Manish','Anjali','Suresh','Kavya','Rahul','Meera',
              'Vikram','Nisha','Arjun','Divya','Rohit','Swathi','Naveen',
              'Priya','Karan','Deepa'],
      'Age':[25,np.nan,30,25,28,22,24,np.nan,27,26,23,29,np.nan,
             25,22,28,24,np.nan,28,26,23,30,27,np.nan,25],
      'Marks':[85,90,np.nan,85,78,95,88,72,np.nan,91,67,84,79,
               np.nan,93,88,76,82,78,np.nan,90,85,71,72,89],
      'City':['Mysuru','Mysuru','Bangaluru','Mysuru',np.nan,'Delhi','Mysuru',
              'Mangalore','Bangaluru',np.nan,'Mysuru','Hubali','Mangalore',
              'Delhi','Bangaluru',np.nan,'Mysuru','Hubali',np.nan,'Delhi',
              'Mangalore','Mysuru','Bangaluru','Mangalore','Hubali']
     }
df = pd.DataFrame(data)
print("Dataset:")
print(df)
```

```python
print("\n Shape of Dataset (rows,columns) : ",df.shape)
print("\n Column Names : ",df.columns.tolist())
print("\n Data Types of each column : ")
print(df.dtypes)
print("\n DataSet info :")
df.info()
```

```python
print("\n Descriptive statistics (numerical columns):")
print(df.describe())
print("Mean:\n",df.mean(numeric_only=True))
print("Median:\n",df.median(numeric_only=True))
print("Standard Deviation:\n",df.std(numeric_only=True))
```

```python
print("\n Missing values in each column: ")
print(df.isnull().sum())
print("\n Total missing values in dataset: ",df.isnull().sum().sum())
```

```python
print("\n Duplicate row in dataset: ")
print(df[df.duplicated()])
print("\n Number of Duplicate Rows:",df.duplicated().sum())
print("\n EDA process completed successfully! ")
```
